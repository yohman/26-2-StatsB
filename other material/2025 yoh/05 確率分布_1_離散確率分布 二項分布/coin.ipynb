# The law of large numbers

We all know that there is 50% chance of getting heads or tails when you toss a coin. So what happens when you toss a coin say, four times? Are you guaranteed to get 2 heads and 2 tails? Well, no, but what if you tossed it 100 times? Will you *approach* 50%?

```python
from ipywidgets import Button, VBox, Output
from IPython.display import clear_output
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Initializing variables
np.random.seed(42)  # Set seed for reproducibility
toss_results = []  # Store toss results (0 or 1)
cumulative_means = []  # Store cumulative means
sample_sizes = []  # Store sample sizes
sample_count = 0  # Initialize sample count
true_probability = 0.5  # True probability of heads

# Output widget for displaying the plot and table
out = Output()

# Function to simulate a coin toss and update the plot and table
def simulate_toss_and_plot(_):
    global toss_results, cumulative_means, sample_sizes, sample_count
    
    toss = np.random.choice([0, 1], p=[1-true_probability, true_probability])  # Random toss
    toss_results.append(toss)  # Append result
    sample_count += 1
    cumulative_mean = np.mean(toss_results)  # Calculate cumulative mean
    cumulative_means.append(cumulative_mean)  # Append cumulative mean
    sample_sizes.append(sample_count)  # Append sample size
    
    # Prepare data for the table
    data = {
        "Sample Size": sample_sizes,
        "Proportion of Heads": cumulative_means,
        "Toss Result": ["Heads" if toss == 1 else "Tails" for toss in toss_results]
    }
    table_df = pd.DataFrame(data).tail(10)  # Show only the last 10 entries for brevity
    
    # Update the plot and table
    with out:
        clear_output(wait=True)
        # Plot
        plt.figure(figsize=(10, 6))
        plt.plot(sample_sizes, cumulative_means, label='Cumulative Mean', color='blue')
        plt.axhline(y=true_probability, color='red', linestyle='--', label='True Probability (0.5)')
        plt.xlim(0, max(100, sample_count + 10))  # Dynamically adjust x-axis
        plt.ylim(0, 1)
        plt.title("Real-Time Law of Large Numbers Simulation")
        plt.xlabel("Sample Size")
        plt.ylabel("Proportion of Heads")
        plt.legend()
        plt.grid(alpha=0.5)
        plt.show()
        
        # Text and Table
        print(f"Current Sample Size: {sample_count}")
        print(f"Current Proportion of Heads: {cumulative_mean:.4f}")
        print("\nLast 10 Results:")
        display(table_df)

# Button to trigger the toss
toss_button = Button(description="Toss Coin")
toss_button.on_click(simulate_toss_and_plot)

# Display the button and output
display(VBox([toss_button, out]))
```

```python
# Initializing variables for dice roll simulation
np.random.seed(42)  # Set seed for reproducibility
dice_results = []  # Store dice roll results (1 to 6)
cumulative_means_dice = []  # Store cumulative means
sample_sizes_dice = []  # Store sample sizes
sample_count_dice = 0  # Initialize sample count
expected_value_dice = 3.5  # Expected value of a fair dice roll

# Output widget for displaying the plot and table
out_dice = Output()

# Function to simulate a dice roll and update the plot and table
def simulate_roll_and_plot(_):
    global dice_results, cumulative_means_dice, sample_sizes_dice, sample_count_dice
    
    roll = np.random.choice([1, 2, 3, 4, 5, 6])  # Random dice roll
    dice_results.append(roll)  # Append result
    sample_count_dice += 1
    cumulative_mean_dice = np.mean(dice_results)  # Calculate cumulative mean
    cumulative_means_dice.append(cumulative_mean_dice)  # Append cumulative mean
    sample_sizes_dice.append(sample_count_dice)  # Append sample size
    
    # Prepare data for the table
    data_dice = {
        "Sample Size": sample_sizes_dice,
        "Cumulative Mean": cumulative_means_dice,
        "Roll Result": dice_results
    }
    table_df_dice = pd.DataFrame(data_dice).tail(10)  # Show only the last 10 entries for brevity
    
    # Update the plot and table
    with out_dice:
        clear_output(wait=True)
        # Plot
        plt.figure(figsize=(10, 6))
        plt.plot(sample_sizes_dice, cumulative_means_dice, label='Cumulative Mean', color='blue')
        plt.axhline(y=expected_value_dice, color='red', linestyle='--', label='Expected Value (3.5)')
        plt.xlim(0, max(100, sample_count_dice + 10))  # Dynamically adjust x-axis
        plt.ylim(1, 6)
        plt.title("Real-Time Law of Large Numbers Simulation (Dice Roll)")
        plt.xlabel("Sample Size")
        plt.ylabel("Cumulative Mean")
        plt.legend()
        plt.grid(alpha=0.5)
        plt.show()
        
        # Text and Table
        print(f"Current Sample Size: {sample_count_dice}")
        print(f"Current Cumulative Mean: {cumulative_mean_dice:.4f}")
        print("\nLast 10 Results:")
        display(table_df_dice)

# Button to trigger the dice roll
roll_button = Button(description="Roll Dice")
roll_button.on_click(simulate_roll_and_plot)

# Display the button and output
display(VBox([roll_button, out_dice]))
```
